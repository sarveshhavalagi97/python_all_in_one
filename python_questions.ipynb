**2) String Compression aaabbc → a3b2c1**

```python
from collections import Counter

text = "aaabbc"

# 1. Let Counter do the counting math automatically
counts = Counter(text)  # Creates: {'a': 3, 'b': 2, 'c': 1}

# 2. Use a normal for loop to build the string
output = ""
for char, count in counts.items():
    output = output+ char + str(count)

print(output)  # Output: a3b2c1
```

**4) Rotate Array by K Positions**

```python
# To rotate an array by K positions to the right, you move the elements at the end of the array to the front.
#For example, if you rotate [1, 2, 3, 4, 5] by K = 2, the output is [4, 5, 1, 2, 3].
```

```python
nums = [1, 2, 3, 4, 5]
k = 2

# 1. Split the list into two parts
back_part = nums[-k:]   # Grabs the last k elements: [4, 5]
front_part = nums[:-k]  # Grabs everything else: [1, 2, 3]

# 2. Put them back together in reverse order
result = back_part + front_part

print(result)  # Output: [4, 5, 1, 2, 3]
```

**5) Find Missing Number in Sequence**

```python
nums = [3, 0, 1]

# 1. Find what the sum SHOULD be if no number was missing
n = len(nums)
expected_sum = (n * (n + 1)) // 2  # Formula: (3 * 4) // 2 = 6

# 2. Find the actual sum of the numbers in the list
actual_sum = sum(nums)             # 3 + 0 + 1 = 4

# 3. The difference is your missing number
missing_number = expected_sum - actual_sum

print(missing_number)  # Output: 2
```

**6) Find Most Frequent Character**

```python
from collections import Counter

text = "hello"

# 1. Count all characters automatically
counts = Counter(text)  # Creates: {'l': 2, 'h': 1, 'e': 1, 'o': 1}

# 2. Get the #1 most common character
# .most_common(1) returns a list with a tuple: [('l', 2)]
most_frequent = counts.most_common(1)[0][0]

print(most_frequent)  # Output: l
```

**8) Group Anagrams**

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
anagram_groups = {}  # A normal dictionary to hold our groups

for word in words:
    # 1. Sort the word just like you did: sorted('eat') -> ['e', 'a', 't']
    # We turn it into a string/tuple so it can be used as a dictionary key
    sorted_key = "".join(sorted(word))  # "aet"
    print(sorted_key)

    # 2. Group them together
    if sorted_key in anagram_groups:
        # If the sorted pattern already exists, add this word to its list
        anagram_groups[sorted_key].append(word)
    else:
        # If it's a new pattern, create a new list with this word
        anagram_groups[sorted_key] = [word]

print(anagram_groups)
# 3. Print the final grouped lists
print(list(anagram_groups.values()))
# Output: [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]
```

**10) Highest Average Marks**

```python
records=[

    "A, S1, 90",
    "B, S1, 80",
    "A, S2, 85",
    "B, S2, 88",
    "C, S1, 70",
    "C, S2, 75",
]

# keep in mind that we have to create like {'A': [90, 85], 'B': [80, 88], 'C': [70, 75]}

n={}

for record in records:
  name, sub, mark= record.split(',')

  mark= int(mark)

  # if name is slready present in the dict then just append the marks to that
  if name in n:
    n[name].append(mark)

  # if name is not present then create new key-value pair
  else:
    n[name]= [mark]

print(n)

highest_avg=0
highest_student=""

for student, values in n.items():

  avg= sum(values)/len(values)

  if avg> highest_avg:
    highest_avg= avg
    highest_student= student

print(highest_student)
```

**12) Expand Encoded String 3a12b6c = aaabbbbbbbbbbbbcc**

```python
# It sees '3'. char.isdigit() is True, so current_number becomes "3".
# Again next it sees 'a'. This is a letter! It converts "3" to the number 3, glues 'a' * 3 ("aaa") to the output, and clears current_number back to "".
```

```python
s = '3a12b6c'

result = ""
number = ""

for ch in s:

    if ch.isdigit():
        number = number + ch

    else:
        result = result + ch * int(number)
        print(number)
        number = ""

print(result)
```

**15. Sort the values: data = [{"suraj": 30, "ram": 20, "shyam": 40, "avi": 50}]**

```python
data = [{"suraj": 30, "ram": 20, "shyam": 40, "avi": 50, "kavi":1}]
inner_dict = data[0]
print("inner_dict", inner_dict)

# 1. Store pairs as (value, key) so Python naturally sorts by the number
flipped_pairs = []
for key, value in inner_dict.items():
    flipped_pairs.append((value, key))

# 2. Sort the pairs (arranges them by value automatically)
flipped_pairs.sort()
print("flipped_pairs",flipped_pairs)

# 3. Rebuild the dictionary in the new sorted order
sorted_dict = {}
for value, key in flipped_pairs:
    sorted_dict[key] = value
print("sorted_dict",sorted_dict)

# 4. Place it back inside a list
sorted_data = [sorted_dict]
print("sorted_data",sorted_data)
```

**16. Sort the keys: data = [{"suraj": 30, "ram": 20, "shyam": 40, "avi": 50}]**

```python
data = [{"suraj": 30, "ram": 20, "shyam": 40, "avi": 50, "kavi": 1}]

# 1. Extract the dictionary from the list
inner_dict = data[0]

# 2. Sort the items (Python naturally sorts pairs alphabetically by their keys)
sorted_items = sorted(inner_dict.items())

# 3. Convert the sorted pairs back into a dictionary and wrap it in a list
sorted_data = [dict(sorted_items)]

print(sorted_data)
# Output: [{'avi': 50, 'kavi': 1, 'ram': 20, 'shyam': 40, 'suraj': 30}]
```

**17. Send only int values to another list including those int which are mentioned in string form l = [2, 5, "name", "7", "class", "9"]**

```python
l = [2, 5, "name", "7", "class", "9"]
int_list = []

for item in l:
    # 1. Check if the item is already an integer
    if isinstance(item, int):
        int_list.append(item)

    # 2. Check if it is a string containing a valid number
    elif isinstance(item, str) and item.isdigit():
        int_list.append(int(item))

print(int_list)
# Output: [2, 5, 7, 9]
```
