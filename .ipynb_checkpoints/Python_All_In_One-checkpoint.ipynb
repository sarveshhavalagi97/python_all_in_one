```python
# Reduce Function

'''
The reduce function in Python is designed to perform a cumulative operation on the items of an iterable, reducing them down to 
a single accumulated result. The reduce function takes a binary function and an iterable as input. It applies the binary 
function cumulatively to the items in the iterable, producing a single accumulated result.

Syntax:
result = reduce(binary_function, iterable, [initializer])

binary_function: A function that takes two arguments and performs a binary operation.
iterable: The sequence of elements to be reduced.
initializer: An optional argument that provides an initial value for the accumulation.'''

# One drawback of list is it considers only 2 elements at a time eg: x,y and not more than that.
# map function can return series of no.'s in o/p but reduce return only single o/p

from functools import reduce

numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print(product)
```

```python
l=[1,2,3,4,5]
def add(a,b):
    return a+b

reduce(add,l) # By its syntax, reduce takes a function (add here) and one iterable variable
```

```python
#Class:
#Class is a blueprint or a template for creating objects. Objects are instances of classes.

class MyClass:
    # attributes
    attribute1 = "Hello"
    attribute2 = 42
    
    # methods
    def my_method(self):
        print("This is a method.")

'''In the example above, MyClass is a simple class with two attributes (attribute1 and attribute2) and one method (my_method). 
The self parameter refers to the instance of the class and is required in method definitions within a class.'''
```

```python
#Polymorphism
'''
Polymorphism is a concept in object-oriented programming that allows objects of different classes to be treated as objects of a 
common base class. It provides a way to use a single interface to represent different types of objects, making code more 
flexible and reusable.'''


class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

# Creating instances
animal_obj = Animal()
dog_obj = Dog()
cat_obj = Cat()

# Using the sound method with different objects
animal_obj.sound()  # Output: Animal makes a sound
dog_obj.sound()     # Output: Dog barks
cat_obj.sound()     # Output: Cat meows
```

```python
# Encapsulation 
'''
It is the idea is to restrict access to some of the object's components and only expose what is necessary. By controlling 
access to data, encapsulation helps in preventing unauthorized access and modification.

self.__make = make =>The double underscores before the attribute names indicate that they are private and should not be 
accessed directly from outside the class.
'''
class Car:
    def __init__(self, make, model):
        self.__make = make  # Private attribute 
        self.model = model   # Public attribute

    def display_info(self):
        print(f"Make: {self.__make}, Model: {self.model}")

# Creating an instance of the Car class
my_car = Car("Toyota", "Camry")

# Accessing public attribute directly
print(my_car.model)  # Output: Camry

# Accessing private attribute directly will result in an error
print(my_car.__make)  # Error
```

```python
class bank_details():
    
    def __init__(self,balance):
        self.__balance= balance
        
    def deposite(self, amount):
        self.__balance= self.__balance+amount
        
    def withdrawl(self, amount):
        if self.__balance>= amount:
            self.__balance = self.__balance-amount
            return True
        else:
            return False
        
    def getbalance(self):
        return self.__balance
```

```python
obj_bank_acc= bank_details(2000)
```

```python
obj_bank_acc.getbalance()
```

```python
obj_bank_acc.deposite(3000)
```

```python
obj_bank_acc.getbalance()
```

```python
obj_bank_acc.withdrawl(2000)
```

```python
# Inheritance
'''Inheritance is a fundamental concept in object-oriented programming (OOP) that allows a class (subclass or derived class) to 
inherit attributes and methods from another class (superclass or base class).'''
```

```python
# Multilable inheritance

class class1():
    def test_class1(self):
        print('this is class1')
        
class class2(class1):
    def test_class2(self):
        print('this is class2')

class class3(class2):
    def test_class3(self):
        print('this is class3')   
```

```python
obj_class3= class3()
```

```python
obj_class3.test_class1() # class3 is able to access class1 as well
```

```python
obj_class3.test_class2()  # class3 is able to access class2 as well
```

```python
# Multiple inheritance

class class1():
    def test_class1(self):
        print('this is class1')
        
class class2():
    def test_class2(self):
        print('this is class2')
        
class class3(class1,class2):
    pass
```

```python
obj_class3= class3()
```

```python
obj_class3.test_class2()  # class3 is able to access class2 as well
```

```python
obj_class3.test_class1() # class3 is able to access class1 as well
```
