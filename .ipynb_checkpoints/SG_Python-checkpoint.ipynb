###### 1.Write a python program to varify the given character is consonent or vowel

```python
def check_vowel_or_consonant(char):
    vowels = "aeiouAEIOU"
    if char.isalpha():  #returns True if all characters in the string are alphabetic
        if char in vowels:
            return f"{char} is a vowel."
        else:
            return f"{char} is a consonant."
```

```python
check_vowel_or_consonant('aw')
```

###### 2.Reverse a number using for loop

```python
def display_numbers_reverse(n):
    for i in range(n, 0, -1):
        print(i)
```

```python
display_numbers_reverse(10)
```

###### Factorial of given number

```python
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result =result * i
    return result
```

```python
factorial_iterative(3)
```

```python
# Use return only not print
def fac(n):
    if n==0 or n==1:
        return 1
    else:
        return n* fac(n-1)
```

```python
fac(3)
```

###### Program to count numbers between 1 to 100 that are divisible by 6

```python
a = 0
# Iterate through numbers from 1 to 100
for i in range(1, 101):
    if i % 6 == 0:
        a= a + 1
print("The count of numbers between 1 and 100 that are divisible by 6 is:", a)
```
